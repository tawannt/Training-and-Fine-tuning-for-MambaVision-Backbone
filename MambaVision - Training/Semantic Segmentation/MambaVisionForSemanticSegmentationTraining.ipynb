## Training with ADE20K

```python
%%writefile train_upernet_mamba.py
import os
import sys

# Ép hệ thống dùng backend Agg TRƯỚC KHI import matplotlib
os.environ["MPLBACKEND"] = "Agg"
import matplotlib
matplotlib.use("Agg")

# Import module chứa kiến trúc backbone MambaVision
sys.path.insert(0, '/content/MambaVision')
import semantic_segmentation.tools.mamba_vision

from mmengine.config import ConfigDict, Config
from mmengine.runner import Runner

def main():
    # =====================================================================
    # 1. CẤU HÌNH DATASET (ADE20K - 150 Classes)
    # =====================================================================
    dataset_type = 'ADE20KDataset'
    data_root = '/content/data/ADEChallengeData2016'
    crop_size = (512, 512)

    data_preprocessor = dict(
        type='SegDataPreProcessor',
        mean=[123.675, 116.28, 103.53],
        std=[58.395, 57.12, 57.375],
        bgr_to_rgb=True,
        pad_val=0,
        seg_pad_val=255,
        size=crop_size
    )

    train_pipeline = [
        dict(type='LoadImageFromFile'),
        dict(type='LoadAnnotations', reduce_zero_label=True),
        dict(type='RandomResize', scale=(2048, 512), ratio_range=(0.5, 2.0), keep_ratio=True),
        dict(type='RandomCrop', crop_size=crop_size, cat_max_ratio=0.75),
        dict(type='RandomFlip', prob=0.5),
        dict(type='PhotoMetricDistortion'),
        dict(type='PackSegInputs')
    ]

    val_pipeline = [
        dict(type='LoadImageFromFile'),
        dict(type='Resize', scale=(2048, 512), keep_ratio=True),
        dict(type='LoadAnnotations', reduce_zero_label=True),
        dict(type='PackSegInputs')
    ]

    train_dataloader = dict(
        batch_size=2,
        num_workers=2,
        persistent_workers=True,
        sampler=dict(type='InfiniteSampler', shuffle=True),
        dataset=dict(
            type=dataset_type,
            data_prefix=dict(img_path='images/training', seg_map_path='annotations/training'),
            pipeline=train_pipeline)
    )

    val_dataloader = dict(
        batch_size=1,
        num_workers=2,
        persistent_workers=True,
        sampler=dict(type='DefaultSampler', shuffle=False),
        dataset=dict(
            type=dataset_type,
            data_prefix=dict(img_path='images/validation', seg_map_path='annotations/validation'),
            pipeline=val_pipeline)
    )

    # =====================================================================
    # 2. CẤU HÌNH MÔ HÌNH (MambaVision + UPerNet Head)
    # =====================================================================
    norm_cfg = dict(type='SyncBN', requires_grad=True)

    model = dict(
        type='EncoderDecoder',
        data_preprocessor=data_preprocessor,
        pretrained=None,
        backbone=dict(
            type='MM_mamba_vision',
            pretrained='/content/mambavision_tiny_pretrained.pth.tar',
            act_cfg=dict(type='GELU'),
            attn_drop_rate=0.0,
            depths=(1, 3, 8, 4),
            dim=80,
            drop_path_rate=0.3,
            drop_rate=0.0,
            embed_dims=96,
            in_dim=32,
            layer_scale=None,
            mlp_ratio=4,
            norm_cfg=dict(requires_grad=True, type='LN'),
            norm_layer='ln2d',
            num_heads=(2, 4, 8, 16),
            out_indices=(0, 1, 2, 3),
            patch_norm=True,
            patch_size=4,
            pretrain_img_size=224,
            qk_scale=None,
            qkv_bias=True,
            strides=(4, 2, 2, 2),
            use_abs_pos_embed=False,
            window_size=(8, 8, 64, 32)
        ),
        decode_head=dict(
            type='UPerHead',
            in_channels=[80, 160, 320, 640],
            in_index=[0, 1, 2, 3],
            pool_scales=(1, 2, 3, 6),
            channels=512,
            dropout_ratio=0.1,
            num_classes=150,
            norm_cfg=norm_cfg,
            align_corners=False,
            loss_decode=dict(type='CrossEntropyLoss', use_sigmoid=False, loss_weight=1.0)
        ),
        auxiliary_head=dict(
            type='FCNHead',
            in_channels=320,
            in_index=2,
            channels=256,
            num_convs=1,
            concat_input=False,
            dropout_ratio=0.1,
            num_classes=150,
            norm_cfg=norm_cfg,
            align_corners=False,
            loss_decode=dict(type='CrossEntropyLoss', use_sigmoid=False, loss_weight=0.4)
        ),
        train_cfg=dict(),
        test_cfg=dict(mode='whole')
    )

    # =====================================================================
    # 3. CHIẾN LƯỢC HUẤN LUYỆN (OPTIMIZER & SCHEDULER)
    # =====================================================================
    optim_wrapper = dict(
        type='AmpOptimWrapper',
        accumulative_counts=8,
        optimizer=dict(type='AdamW', lr=5e-05, betas=(0.9, 0.999), weight_decay=0.01),
        paramwise_cfg=dict(custom_keys=dict(norm=dict(decay_mult=0.0)))
    )

    param_scheduler = [
        dict(type='LinearLR', start_factor=1e-06, by_epoch=False, begin=0, end=1500),
        dict(type='PolyLR', eta_min=0.0, power=1.0, begin=1500, end=60000, by_epoch=False)
    ]

    # ---> CẬP NHẬT CHIẾN LƯỢC MỚI: Rút gọn xuống 20k steps, Val mỗi 1000 steps <---
    train_cfg = dict(
        type='IterBasedTrainLoop',
        max_iters=60000,
        val_interval=6000
    )
    val_cfg = dict(type='ValLoop')
    test_cfg = dict(type='TestLoop')

    val_evaluator = dict(type='IoUMetric', iou_metrics=['mIoU'])
    test_evaluator = val_evaluator

    default_hooks = dict(
        timer=dict(type='IterTimerHook'),
        logger=dict(type='LoggerHook', interval=50, log_metric_by_epoch=False),
        param_scheduler=dict(type='ParamSchedulerHook'),
        # ---> CẬP NHẬT CHECKPOINT HOOK: Lưu mỗi 1000 steps, giữ 3 file, và có file Best <---
        checkpoint=dict(
            type='CheckpointHook',
            by_epoch=False,
            interval=1000,
            max_keep_ckpts=3,
            save_best='mIoU',
            rule='greater'
        ),
        sampler_seed=dict(type='DistSamplerSeedHook'),
        visualization=dict(type='SegVisualizationHook')
    )

    # =====================================================================
    # 4. GÓP CẤU HÌNH VÀ CHẠY
    # =====================================================================
    cfg_dict = dict(
        default_scope='mmseg',
        env_cfg=dict(cudnn_benchmark=True, mp_cfg=dict(mp_start_method='fork', opencv_num_threads=0), dist_cfg=dict(backend='nccl')),
        vis_backends=[dict(type='LocalVisBackend')],
        visualizer=dict(type='SegLocalVisualizer', vis_backends=[dict(type='LocalVisBackend')], name='visualizer'),
        log_processor=dict(by_epoch=False),
        log_level='INFO',

        # ---> RESUME TRAINING: Nạp trọng số từ checkpoint cũ của ADE20K <---
        # Sửa tên 'iter_XXXX.pth' theo đúng file mà bạn đang có trong Drive
        load_from='/content/drive/MyDrive/Pattern Recognition/Source Code/MambaVision - Training/Semantic Segmentation/work_dirs/iter_55000.pth',
        resume=True,

        work_dir='/content/drive/MyDrive/Pattern Recognition/Source Code/MambaVision - Training/Semantic Segmentation/work_dirs',

        data_preprocessor=data_preprocessor,
        model=model,
        train_dataloader=train_dataloader,
        val_dataloader=val_dataloader,
        test_dataloader=val_dataloader,
        train_cfg=train_cfg,
        val_cfg=val_cfg,
        test_cfg=test_cfg,
        optim_wrapper=optim_wrapper,
        param_scheduler=param_scheduler,
        default_hooks=default_hooks,
        val_evaluator=val_evaluator,
        test_evaluator=test_evaluator
    )

    cfg = Config(cfg_dict)

    for dl in [cfg.train_dataloader, cfg.val_dataloader, cfg.test_dataloader]:
        dl['dataset']['data_root'] = data_root

    print("🚀 Bắt đầu huấn luyện MambaVision + UPerNet (ADE20K)")
    runner = Runner.from_cfg(cfg)
    runner.train()

if __name__ == '__main__':
    main()
```

# Testing

## Inference Test

```python
%%writefile inference_ade20k.py
import os
import sys
import cv2
import torch
import numpy as np

if "MPLBACKEND" in os.environ:
    del os.environ["MPLBACKEND"]

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

sys.path.insert(0, '/content/MambaVision')
import semantic_segmentation.tools.mamba_vision

from mmengine.config import Config
from mmseg.apis import init_model, inference_model
from mmseg.utils import get_classes, get_palette

def main():
    base_config_path = '/content/MambaVision/semantic_segmentation/configs/mamba_vision/mamba_vision_160k_ade20k-512x512_tiny.py'
    checkpoint_path = '/content/drive/MyDrive/Pattern Recognition/Source Code/MambaVision - Training/Semantic Segmentation/work_dirs/best_mIoU_iter_60000.pth'
    img_path = '/content/data/ADEChallengeData2016/images/validation/ADE_val_00000150.jpg'

    if not os.path.exists(checkpoint_path):
        print(f"❌ Không tìm thấy file trọng số tại: {checkpoint_path}")
        return

    cfg = Config.fromfile(base_config_path)

    num_classes = 150
    cfg.model.decode_head.num_classes = num_classes
    cfg.model.auxiliary_head.num_classes = num_classes

    if 'pretrained' in cfg.model.backbone:
        cfg.model.backbone.pretrained = None

    print("⏳ Đang khởi tạo mô hình MambaVision...")
    model = init_model(cfg, checkpoint_path, device='cuda:0')

    ade_classes = get_classes('ade20k')
    ade_palette = get_palette('ade20k')

    print(f"🚀 Đang inference ảnh: {os.path.basename(img_path)}")
    result = inference_model(model, img_path)

    pred_mask = result.pred_sem_seg.data[0].cpu().numpy()

    plt.figure(figsize=(16, 8))

    plt.subplot(1, 2, 1)
    original_img = cv2.imread(img_path)
    plt.imshow(cv2.cvtColor(original_img, cv2.COLOR_BGR2RGB))
    plt.title(f"Original Image\n({os.path.basename(img_path)})", fontweight='bold')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    color_mask = np.zeros((pred_mask.shape[0], pred_mask.shape[1], 3), dtype=np.uint8)
    for label, color in enumerate(ade_palette):
        color_mask[pred_mask == label] = color

    plt.imshow(color_mask)
    plt.title("MambaVision Prediction (ADE20K)", fontweight='bold')
    plt.axis('off')

    plt.tight_layout()
    output_name = 'ade20k_result.png'
    plt.savefig(output_name, dpi=300)
    print(f"✅ Thành công! Kết quả đã lưu tại: {output_name}")

if __name__ == '__main__':
    main()
```
