## Align LaPa dataset

```python
import os
import cv2
import numpy as np
import glob

def process_lapa_alignment_farl_style(src_dir, dst_dir, target_size=512, margin_ratio=0.25):
    """
    Tiền xử lý tập LaPa chuẩn SOTA (Lấy cảm hứng từ FaRL refinebb):
    Dùng Landmark -> Tính Box Vuông -> Thêm Margin -> Chêm Padding (chống méo) -> Resize.
    """
    splits = ['train', 'val', 'test']

    for split in splits:
        print(f"\nĐang xử lý tập: {split.upper()}...")

        # Tạo cấu trúc thư mục đích
        os.makedirs(os.path.join(dst_dir, split, 'images'), exist_ok=True)
        os.makedirs(os.path.join(dst_dir, split, 'labels'), exist_ok=True)

        img_paths = glob.glob(os.path.join(src_dir, split, 'images', '*.jpg'))
        total_imgs = len(img_paths)

        for idx, img_path in enumerate(img_paths):
            base_name = os.path.basename(img_path)
            name_no_ext = os.path.splitext(base_name)[0]

            label_path = os.path.join(src_dir, split, 'labels', name_no_ext + '.png')
            landmark_path = os.path.join(src_dir, split, 'landmarks', name_no_ext + '.txt')

            # Bỏ qua nếu thiếu file mask hoặc landmark
            if not os.path.exists(label_path) or not os.path.exists(landmark_path):
                continue

            # 1. Đọc Ảnh (RGB/BGR) và Mask (Grayscale)
            img = cv2.imread(img_path)
            mask = cv2.imread(label_path, cv2.IMREAD_GRAYSCALE)

            # 2. Đọc 106 điểm Landmarks (Chuẩn code Microsoft FaRL)
            try:
                raw_lmk_data = [float(v) for v in open(landmark_path, 'r').read().split()]
                assert raw_lmk_data[0] == 106 and len(raw_lmk_data) == 106 * 2 + 1
                landmarks = np.reshape(np.array(raw_lmk_data[1:], dtype=np.float32), [106, 2])
            except Exception as e:
                print(f"Lỗi đọc landmark file {landmark_path}: {e}")
                continue

            # 3. Tính Bounding Box từ 106 điểm
            x_min, y_min = np.min(landmarks, axis=0)
            x_max, y_max = np.max(landmarks, axis=0)

            w = x_max - x_min
            h = y_max - y_min

            # 4. Thuật toán RefineBB (Ép khung vuông & Thêm lề)
            side = max(w, h)
            x_center = x_min + w / 2.0
            y_center = y_min + h / 2.0

            new_side = int(side * (1 + margin_ratio))

            # Tọa độ hộp bao lý tưởng (Có thể bị âm hoặc vượt quá kích thước ảnh)
            x1_raw = int(x_center - new_side / 2)
            y1_raw = int(y_center - new_side / 2)
            x2_raw = x1_raw + new_side
            y2_raw = y1_raw + new_side

            # 5. Tính toán Padding (Lề viền đen) cần chêm vào nếu hộp bị tràn
            pad_left = max(0, -x1_raw)
            pad_top = max(0, -y1_raw)
            pad_right = max(0, x2_raw - img.shape[1])
            pad_bottom = max(0, y2_raw - img.shape[0])

            # Tọa độ an toàn để cắt xén trên bức ảnh gốc thực tế
            x1_safe = max(0, x1_raw)
            y1_safe = max(0, y1_raw)
            x2_safe = min(img.shape[1], x2_raw)
            y2_safe = min(img.shape[0], y2_raw)

            # Cắt ảnh và mask
            cropped_img_safe = img[y1_safe:y2_safe, x1_safe:x2_safe]
            cropped_mask_safe = mask[y1_safe:y2_safe, x1_safe:x2_safe]

            # Kiểm tra tránh lỗi crop rỗng
            if cropped_img_safe.size == 0 or cropped_mask_safe.size == 0:
                continue

            # 6. Kỹ thuật Padding cốt lõi: Bù viền đen để tạo hình vuông hoàn hảo (1:1)
            # Ảnh RGB bù màu đen [0,0,0], Mask 1 kênh bù giá trị 0 (Background)
            square_img = cv2.copyMakeBorder(cropped_img_safe, pad_top, pad_bottom, pad_left, pad_right, cv2.BORDER_CONSTANT, value=[0, 0, 0])
            square_mask = cv2.copyMakeBorder(cropped_mask_safe, pad_top, pad_bottom, pad_left, pad_right, cv2.BORDER_CONSTANT, value=0)

            # 7. Phóng to/Thu nhỏ về kích thước đầu vào của Model (512x512)
            # QUAN TRỌNG: Ảnh dùng Linear cho mượt, Mask dùng Nearest để bảo toàn các số nguyên 0-11
            final_img = cv2.resize(square_img, (target_size, target_size), interpolation=cv2.INTER_LINEAR)
            final_mask = cv2.resize(square_mask, (target_size, target_size), interpolation=cv2.INTER_NEAREST)

            # 8. Lưu thành phẩm
            save_img_path = os.path.join(dst_dir, split, 'images', base_name)
            save_mask_path = os.path.join(dst_dir, split, 'labels', name_no_ext + '.png')

            cv2.imwrite(save_img_path, final_img)
            cv2.imwrite(save_mask_path, final_mask)

            # In tiến độ
            if (idx + 1) % 1000 == 0 or (idx + 1) == total_imgs:
                print(f"  [+] Đã xử lý {idx + 1}/{total_imgs} ảnh...")


# ==========================================
# KHU VỰC CHẠY (Ông nhớ thay đổi đường dẫn phù hợp với Google Drive của ông)
# ==========================================
# Ví dụ:
# SRC_DIR = '/content/drive/MyDrive/LaPa_Raw'
# DST_DIR = '/content/drive/MyDrive/LaPa_Aligned_512'

SRC_DIR = '/content/data/LaPa'  # Trỏ vào thư mục chứa dataset gốc của ông
DST_DIR = '/content/data/LaPa_Aligned_512'  # Thư mục mới sẽ được tạo ra

process_lapa_alignment_farl_style(SRC_DIR, DST_DIR, target_size=512, margin_ratio=0.25)
```

## Comparing LaPa to LaPa Aligned

### Minh hoạ

```python
import os
import cv2
import numpy as np
import matplotlib
matplotlib.use("Agg") # Chạy trên Colab không bị lỗi hiển thị
import matplotlib.pyplot as plt
import matplotlib.patches as patches

def visualize_face_alignment(img_path, landmark_path, output_path, margin_ratio=0.25):
    # 1. Đọc ảnh gốc
    if not os.path.exists(img_path) or not os.path.exists(landmark_path):
        print("Lỗi: Không tìm thấy file ảnh hoặc file landmark!")
        return

    img = cv2.imread(img_path)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # 2. Đọc 106 điểm Landmarks
    raw_lmk_data = [float(v) for v in open(landmark_path, 'r').read().split()]
    landmarks = np.reshape(np.array(raw_lmk_data[1:], dtype=np.float32), [106, 2])

    # ==========================================
    # BƯỚC 1: TÍNH TOÁN HỘP GIỚI HẠN THÔ (MÀU ĐỎ)
    # ==========================================
    x_min, y_min = np.min(landmarks, axis=0)
    x_max, y_max = np.max(landmarks, axis=0)

    w = x_max - x_min
    h = y_max - y_min

    # ==========================================
    # BƯỚC 2: TỐI ƯU HÓA HỘP HÌNH VUÔNG (MÀU XANH)
    # ==========================================
    side = max(w, h)
    x_c = x_min + w / 2.0
    y_c = y_min + h / 2.0

    new_side = int(side * (1 + margin_ratio))

    x1 = int(x_c - new_side / 2)
    y1 = int(y_c - new_side / 2)
    x2 = x1 + new_side
    y2 = y1 + new_side

    # ==========================================
    # VẼ TRỰC QUAN LÊN ẢNH BẰNG MATPLOTLIB
    # ==========================================
    fig, ax = plt.subplots(1, 1, figsize=(10, 10))
    ax.imshow(img_rgb)

    # Vẽ 106 điểm Landmarks (Chấm vàng)
    ax.scatter(landmarks[:, 0], landmarks[:, 1], c='yellow', s=8, edgecolors='black', linewidth=0.5, label='106 Facial Landmarks')

    # Vẽ Initial Bounding Box (Hình chữ nhật Đỏ, nét đứt)
    rect_raw = patches.Rectangle((x_min, y_min), w, h, linewidth=2, edgecolor='red', facecolor='none', linestyle='--', label='Initial BBox (Tight)')
    ax.add_patch(rect_raw)

    # Vẽ Refined Bounding Box (Hình vuông Xanh lá, nét liền)
    rect_refined = patches.Rectangle((x1, y1), new_side, new_side, linewidth=3, edgecolor='lime', facecolor='none', label=f'Refined BBox (Square, Margin={margin_ratio*100}%)')
    ax.add_patch(rect_refined)

    # Cấu hình thẩm mỹ
    ax.set_title("Bounding Box Refinement", fontsize=16, pad=15)
    ax.axis('off') # Tắt hiển thị trục tọa độ

    # Thêm chú thích
    ax.legend(loc='lower center', bbox_to_anchor=(0.5, -0.1), ncol=3, frameon=False, fontsize=12)

    plt.tight_layout()
    plt.savefig(output_path, dpi=300, bbox_inches='tight')
    print(f"Đã vẽ thành công! Vui lòng xem ảnh tại: {output_path}")


def main():
    # CHÚ Ý: Đổi đường dẫn này tới một bức ảnh bất kỳ trong tập LaPa của ông
    sample_img_path = '/content/data/LaPa/test/images/10036529425_0.jpg'
    sample_lmk_path = '/content/data/LaPa/test/landmarks/10036529425_0.txt'

    # Nơi lưu ảnh xuất ra để dán vào Word/LaTeX
    output_image = 'visualize_bbox_refinement.png'

    visualize_face_alignment(sample_img_path, sample_lmk_path, output_image, margin_ratio=0.25)

if __name__ == '__main__':
    main()
```

### Thống kê

```python
import os
import cv2
import glob
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

plt.rcParams.update({
    "font.family": "serif",
    "font.weight": "normal",
    "axes.titleweight": "normal",
    "axes.labelweight": "normal",
    "font.size": 12,
    "axes.titlesize": 14,
    "axes.labelsize": 12,
    "legend.fontsize": 11,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "axes.spines.top": False,
    "axes.spines.right": False
})

def analyze_dataset_noise(mask_dir, dataset_name):
    """Phân tích mức độ nhiễu và vị trí khuôn mặt dựa trên Mask."""
    mask_paths = glob.glob(os.path.join(mask_dir, '*.png'))

    snr_list = []
    center_x_list = []
    center_y_list = []

    print(f"[*] Analyzing {dataset_name} ({len(mask_paths)} images)...")

    for path in mask_paths:
        mask = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
        if mask is None:
            continue

        h, w = mask.shape

        # 1. Tính Tỷ lệ Tín hiệu / Nhiễu (SNR)
        face_pixels = np.sum(mask > 0)
        bg_pixels = np.sum(mask == 0)

        if face_pixels > 0 and bg_pixels > 0:
            snr = face_pixels / bg_pixels
            snr_list.append(snr)

            # 2. Tính Tọa độ Trọng tâm Khuôn mặt
            y_coords, x_coords = np.where(mask > 0)

            mean_x = np.mean(x_coords) / w
            mean_y = np.mean(y_coords) / h

            center_x_list.append(mean_x)
            center_y_list.append(mean_y)

    metrics = {
        "mean_snr": np.mean(snr_list),
        "std_snr": np.std(snr_list),
        "var_x": np.var(center_x_list),
        "var_y": np.var(center_y_list)
    }

    return metrics, snr_list, center_x_list, center_y_list


def main():
    raw_val_masks = '/content/data/LaPa/train/labels'
    aligned_val_masks = '/content/data/LaPa_Aligned_512/train/labels'

    # Chạy phân tích
    raw_metrics, raw_snr, raw_cx, raw_cy = analyze_dataset_noise(raw_val_masks, "LaPa_Raw")
    align_metrics, align_snr, align_cx, align_cy = analyze_dataset_noise(aligned_val_masks, "LaPa_Aligned")

    # In báo cáo Console
    print("\n" + "="*60)
    print("ALIGNMENT NOISE REDUCTION REPORT".center(60))
    print("="*60)
    print(f"{'Metric':<25} | {'Raw Dataset':<15} | {'Aligned Dataset':<15}")
    print("-" * 60)
    print(f"{'1. Mean SNR':<25} | {raw_metrics['mean_snr']:.4f}          | {align_metrics['mean_snr']:.4f}")
    print(f"{'2. Std SNR':<25} | {raw_metrics['std_snr']:.4f}          | {align_metrics['std_snr']:.4f}")
    print(f"{'3. X-Center Variance':<25} | {raw_metrics['var_x']:.5f}         | {align_metrics['var_x']:.5f}")
    print(f"{'4. Y-Center Variance':<25} | {raw_metrics['var_y']:.5f}         | {align_metrics['var_y']:.5f}")

    fig, axes = plt.subplots(1, 2, figsize=(14, 6))

    # Màu sắc học thuật: Đỏ trầm và Xanh dịu
    color_raw = '#d62728'
    color_aligned = '#1f77b4'

    # --- Subplot 1: Histogram của SNR ---
    # Thêm edgecolor='white' để tách các cột nhìn sạch hơn
    axes[0].hist(raw_snr, bins=50, alpha=0.6, label='Raw Dataset',
                 color=color_raw, density=True, edgecolor='white', linewidth=0.5)
    axes[0].hist(align_snr, bins=50, alpha=0.6, label='Aligned Dataset',
                 color=color_aligned, density=True, edgecolor='white', linewidth=0.5)

    axes[0].set_title('Signal-to-Noise Ratio (SNR) Distribution')
    axes[0].set_xlabel('Face-to-Background Area Ratio')
    axes[0].set_ylabel('Density')
    axes[0].legend(frameon=False) # Bỏ viền khung chú thích
    axes[0].grid(axis='y', linestyle='--', alpha=0.5) # Chỉ vẽ lưới ngang

    # --- Subplot 2: Scatter của Center of Mass ---
    # Dùng edgecolors='none' để các chấm không bị viền đen làm rối mắt
    axes[1].scatter(raw_cx, raw_cy, alpha=0.4, label='Raw Dataset',
                    color=color_raw, s=15, edgecolors='none')
    axes[1].scatter(align_cx, align_cy, alpha=0.5, label='Aligned Dataset',
                    color=color_aligned, s=15, edgecolors='none')

    axes[1].set_xlim(0, 1)
    axes[1].set_ylim(1, 0)
    axes[1].set_title('Normalized Face Center of Mass')
    axes[1].set_xlabel('Normalized X Coordinate')
    axes[1].set_ylabel('Normalized Y Coordinate')
    axes[1].legend(frameon=False)
    axes[1].grid(True, linestyle='--', alpha=0.5) # Lưới ô vuông mờ

    plt.tight_layout()
    plt.savefig('alignment_proof_paper.png', dpi=300, bbox_inches='tight')
    print("\nSaved academic-style plot to: alignment_proof_paper.png")

if __name__ == '__main__':
    main()
```

```python
%%writefile extract.py
import torch
import os
import shutil

input_path = 'mambavision_base_1k.pth.tar'
tmp_output = '/content/mamba_backbone_refined.pth'

# Tôi vừa fix lại lỗi chính tả (MambaVsion -> MambaVision) ở đường dẫn của ông
final_drive_path = '/content/drive/MyDrive/MambaVision/Mask2Former/mamba_backbone_refined.pth'

print(f"--- Đang đọc dữ liệu từ {input_path} ---")

try:
    checkpoint = torch.load(input_path, map_location='cpu')

    # 1. SỬA LỖI CHÍ MẠNG Ở ĐÂY: Tìm đúng chìa khóa chứa trọng số
    if 'state_dict' in checkpoint:
        raw_dict = checkpoint['state_dict']
    elif 'model' in checkpoint:
        raw_dict = checkpoint['model']
    else:
        raw_dict = checkpoint

    refined_dict = {}

    # 2. Xử lý từng layer
    for k, v in raw_dict.items():
        # Xóa các tiền tố thừa (module.) nếu có
        clean_key = k.replace('module.', '').replace('backbone.', '')

        # Bỏ qua lớp head phân loại của ImageNet (Không dùng cho Segmentation)
        if any(x in clean_key for x in ['head', 'norm.weight', 'norm.bias']):
            continue

        # Đóng gói lại chuẩn chỉ cho MMSegmentation
        new_key = f'backbone.{clean_key}'
        refined_dict[new_key] = v

    # 3. Lưu file sạch
    torch.save(refined_dict, tmp_output)
    print(f"✅ Đã xử lý xong {len(refined_dict)} layers chuẩn.")

    # Copy sang Drive
    os.makedirs(os.path.dirname(final_drive_path), exist_ok=True)
    shutil.copy(tmp_output, final_drive_path)

    print(f"🚀 THÀNH CÔNG! File CHUẨN đã nằm tại: {final_drive_path}")

except Exception as e:
    print(f"❌ LỖI: {e}")
```

```python
%%writefile train_phase2.py
import os
import sys
import torch

os.environ["MPLBACKEND"] = "Agg"
import matplotlib
matplotlib.use("Agg")

# Trỏ vào repo MambaVision của ông
sys.path.insert(0, '/content/MambaVision')
import semantic_segmentation.tools.mamba_vision

from mmengine.config import Config
from mmengine.runner import Runner

def clean_pipeline(pipeline):
    new_pipeline = []

    # Kích thước mục tiêu cho LaPa Aligned
    target_size = (512, 512)

    for t in pipeline:
        p_type = t.get('type', '')

        # 1. Giữ nguyên setup nhãn
        if p_type == 'LoadAnnotations':
            t['reduce_zero_label'] = False
            new_pipeline.append(t)
            continue

        # 2. CHẶN ĐỨNG mọi thể loại Resize ngẫu nhiên của Mask2Former
        if p_type in ['RandomFlip', 'RandomCrop', 'RandomResize', 'RandomChoiceResize']:
            continue

        # 3. THAY THẾ Resize cũ bằng Resize cứng
        if p_type == 'Resize':
            t['scale'] = target_size
            t['keep_ratio'] = False # Ép về đúng 512x512, không giữ tỉ lệ
            new_pipeline.append(t)
        else:
            new_pipeline.append(t)

    # 4. CHỐT CHẶN CUỐI: Nhét Resize vào nếu bị thiếu
    has_resize = any(p.get('type') == 'Resize' for p in new_pipeline)
    if not has_resize:
        # Chèn vào vị trí an toàn (sau khi load ảnh và nhãn xong)
        new_pipeline.insert(2, dict(type='Resize', scale=target_size, keep_ratio=False))

    return new_pipeline

def main():
    # 1. LOAD BASE CONFIG MASK2FORMER
    base_config_path = '/content/MambaVision/semantic_segmentation/configs/mask2former/mask2former_swin-s_8xb2-160k_ade20k-512x512.py'

    try:
        cfg = Config.fromfile(base_config_path)
    except FileNotFoundError:
        print("LỖI: Không tìm thấy file base Mask2Former. Hãy chạy lệnh git clone mmseg đã nhé!")
        return

    # 2. ĐỘ KIẾN TRÚC: THAY SWIN BẰNG MAMBAVISION
    cfg.model.backbone = dict(
        type='MM_mamba_vision',
        depths=(3, 3, 10, 5),
        embed_dims=96,
        dim=128,
        in_dim=64,
        mlp_ratio=4.0,
        num_heads=(2, 4, 8, 16),
        window_size=(8, 8, 64, 32),
        drop_path_rate=0.4,
        layer_scale=1e-05,
        norm_layer='ln2d',
        out_indices=(0, 1, 2, 3),
        init_cfg=None # QUAN TRỌNG: Xóa sạch init_cfg để không nạp nhầm weights Swin
    )

    # 3. ĐỒNG BỘ KÊNH (CHANNELS) CHO MASK2FORMER HEAD & PIXEL DECODER
    mamba_channels = [128, 256, 512, 1024]
    cfg.model.decode_head.in_channels = mamba_channels

    # Bắt buộc phải cập nhật cả in_channels của pixel_decoder
    if 'pixel_decoder' in cfg.model.decode_head:
        cfg.model.decode_head.pixel_decoder.in_channels = mamba_channels

    cfg.model.decode_head.num_classes = 11  # LaPa 11 classes

    # VÁ LỖI TENSOR 151: Cập nhật lại mảng class_weight cho hàm Loss
    # 11 số 1.0 cho các class khuôn mặt, và 1 số 0.1 cho class ẩn "no_object"
    cfg.model.decode_head.loss_cls.class_weight = [1.0] * 11 + [0.1]

    # 4. SETUP DATASET LAPA
    cfg.dataset_type = 'BaseSegDataset'
    cfg.data_root = '/content/data/LaPa_Aligned_512'

    classes = ('background', 'skin', 'left_eyebrow', 'right_eyebrow', 'left_eye',
               'right_eye', 'nose', 'upper_lip', 'inner_mouth', 'lower_lip', 'hair')
    metainfo = dict(classes=classes)

    for dl in [cfg.train_dataloader, cfg.val_dataloader, cfg.test_dataloader]:
        dl['dataset']['type'] = cfg.dataset_type
        dl['dataset']['data_root'] = cfg.data_root
        dl['dataset']['metainfo'] = metainfo
        dl['dataset']['reduce_zero_label'] = False
        if 'pipeline' in dl['dataset']:
            dl['dataset']['pipeline'] = clean_pipeline(dl['dataset']['pipeline'])

    cfg.train_dataloader.dataset.data_prefix = dict(img_path='train/images', seg_map_path='train/labels')
    cfg.val_dataloader.dataset.data_prefix = dict(img_path='val/images', seg_map_path='val/labels')

    # 5. CHIẾN THUẬT VRAM CHO TESLA T4
    cfg.train_dataloader.batch_size = 2
    cfg.optim_wrapper.accumulative_counts = 8 # Batch ảo = 16

# ==========================================================
    # 1. NẠP WEIGHTS TỪ ĐỈNH CAO PHASE 1
    # ==========================================================
    cfg.load_from = '/content/drive/MyDrive/MambaVision/Mask2Former/work_dirs2/phase2/iter_49000.pth'

    # ÉP BUỘC RESET TRẠNG THÁI (Lấy Weights, bỏ Optimizer cũ)
    cfg.resume = True

    # ==========================================================
    # 2. CHIẾN THUẬT LR MỚI (Hạ Base, Tăng Backbone)
    # ==========================================================
    # Giảm Base LR đi một nửa (từ 1e-4 xuống 5e-5).
    # Vì đầu Mask2Former đã đạt 74%, LR to quá sẽ làm hỏng weights.
    # cfg.optim_wrapper.optimizer.lr = 5e-5

    cfg.optim_wrapper.paramwise_cfg = dict(
        custom_keys={
            'backbone': dict(lr_mult=0.1, decay_mult=1.0), # <-- MỞ VAN LÊN 0.1
            'query_embed': dict(lr_mult=1.0, decay_mult=0.0),
            'query_feat': dict(lr_mult=1.0, decay_mult=0.0),
            'level_embed': dict(lr_mult=1.0, decay_mult=0.0)
        }
    )

    # ==========================================================
    # 3. VẼ LẠI ĐƯỜNG CONG POLY-LR MỚI
    # ==========================================================
    cfg.param_scheduler = [
        dict(
            type='PolyLR',
            eta_min=0,
            power=0.9,
            begin=0,
            end=60000,     # Vẽ đường dốc kéo dài 60k vòng
            by_epoch=False)
    ]

    # ==========================================================
    # 4. CHU KỲ HUẤN LUYỆN (60k Iters)
    # ==========================================================
    cfg.train_cfg.max_iters = 60000
    cfg.train_cfg.val_interval = 6000

    cfg.default_hooks.checkpoint = dict(
        type='CheckpointHook',
        by_epoch=False,
        interval=1000,
        max_keep_ckpts=2,
        save_best='mIoU',
        rule='greater'
    )


    cfg.work_dir = '/content/drive/MyDrive/MambaVision/Mask2Former/work_dirs2/phase2'

    runner = Runner.from_cfg(cfg)
    runner.train()

if __name__ == '__main__':
    main()
```

```python
%%writefile inference.py
import os
import sys
import cv2
import torch
import numpy as np

# Xóa backend Colab để tránh lỗi hiển thị ngầm
if "MPLBACKEND" in os.environ:
    del os.environ["MPLBACKEND"]

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

# Cắm repo MambaVision vào đường dẫn hệ thống
sys.path.insert(0, '/content/MambaVision')
import semantic_segmentation.tools.mamba_vision

from mmengine.config import Config
from mmseg.apis import init_model, inference_model

def clean_test_pipeline(pipeline):
    """
    Hàm này cực kỳ quan trọng cho Inference!
    Nó ép bức ảnh đầu vào phải trải qua quy trình tiền xử lý
    Y HỆT như lúc ông validation (Resize thẳng về 512x512, không bóp méo).
    """
    new_pipeline = []
    target_size = (512, 512)

    for t in pipeline:
        p_type = t.get('type', '')
        # Loại bỏ LoadAnnotations vì lúc test thực tế không có nhãn
        if p_type == 'LoadAnnotations':
            continue

        # Sửa LoadImageFromFile thành loại không hỗ trợ các tính năng dư thừa
        if p_type == 'LoadImageFromFile':
            new_pipeline.append(t)
            continue

        # Ép Resize cứng, không giữ tỷ lệ
        if p_type == 'Resize':
            t['scale'] = target_size
            t['keep_ratio'] = False
            new_pipeline.append(t)
            continue

        new_pipeline.append(t)

    # Nếu thiếu Resize, chèn vào sau LoadImage
    has_resize = any(p.get('type') == 'Resize' for p in new_pipeline)
    if not has_resize:
        new_pipeline.insert(1, dict(type='Resize', scale=target_size, keep_ratio=False))

    return new_pipeline

def main():
    print("🚀 Bắt đầu nạp mô hình Mask2Former - MambaVision...")

    # 1. LOAD CONFIG ĐÚNG NHƯ LÚC TRAIN (MASK2FORMER)
    base_config_path = '/content/MambaVision/semantic_segmentation/configs/mask2former/mask2former_swin-s_8xb2-160k_ade20k-512x512.py'

    try:
        cfg = Config.fromfile(base_config_path)
    except FileNotFoundError:
        print(f"Lỗi: Không tìm thấy file {base_config_path}")
        return

    # 2. XÂY DỰNG LẠI KIẾN TRÚC MÔ HÌNH (Khớp 100% với file train_phase2.py)
    cfg.model.backbone = dict(
        type='MM_mamba_vision',
        depths=(3, 3, 10, 5), embed_dims=96, dim=128, in_dim=64,
        mlp_ratio=4.0, num_heads=(2, 4, 8, 16), window_size=(8, 8, 64, 32),
        drop_path_rate=0.4, layer_scale=1e-05, norm_layer='ln2d',
        out_indices=(0, 1, 2, 3), init_cfg=None
    )

    mamba_channels = [128, 256, 512, 1024]
    cfg.model.decode_head.in_channels = mamba_channels
    if 'pixel_decoder' in cfg.model.decode_head:
        cfg.model.decode_head.pixel_decoder.in_channels = mamba_channels

    cfg.model.decode_head.num_classes = 11  # LaPa 11 classes

    # Fix logic Pipeline cho Inference
    if 'test_pipeline' in cfg:
        cfg.test_pipeline = clean_test_pipeline(cfg.test_pipeline)
    elif 'test_dataloader' in cfg and 'dataset' in cfg.test_dataloader and 'pipeline' in cfg.test_dataloader.dataset:
        cfg.test_pipeline = clean_test_pipeline(cfg.test_dataloader.dataset.pipeline)
    else:
        # Nếu không tìm thấy, tự tạo 1 pipeline chuẩn
        cfg.test_pipeline = [
            dict(type='LoadImageFromFile'),
            dict(type='Resize', scale=(512, 512), keep_ratio=False),
            dict(type='PackSegInputs')
        ]

    # 3. NẠP TRỌNG SỐ (Sửa lại đường dẫn của ông)
    # Ví dụ: đường dẫn này phải trỏ tới file Best mIoU của Phase 2
    checkpoint_path = '/content/drive/MyDrive/Pattern Recognition/Source Code/MambaVision - Finetuning/Data Engineer/best_mIoU_iter_48000.pth'

    if not os.path.exists(checkpoint_path):
        print(f"Lỗi: Không tìm thấy weights tại: {checkpoint_path}")
        return

    # Khởi tạo mô hình
    model = init_model(cfg, checkpoint_path, device='cuda:0')

    # 4. THỰC THI INFERENCE
    # Chọn 1 tấm ảnh trong tập Aligned (đã cắt vuông)
    img_path = '/content/data/LaPa_Aligned_512/test/images/10036529425_0.jpg'

    if not os.path.exists(img_path):
        print(f"Lỗi: Không tìm thấy ảnh tại: {img_path}")
        return

    print("🧠 Đang phân tích ảnh...")
    try:
        # Lệnh inference_model của MMSegmentation
        result = inference_model(model, img_path)
    except Exception as e:
        print(f"Lỗi khi chạy Inference: {e}")
        return

    # 5. XỬ LÝ VÀ VẼ HÌNH
    # Lấy ma trận kết quả (11 classes)
    pred_mask = result.pred_sem_seg.data[0].cpu().numpy()

    # Tạo figure
    plt.figure(figsize=(15, 7))

    # ---- Hình 1: Ảnh gốc ----
    plt.subplot(1, 3, 1)
    original_img = cv2.imread(img_path)
    original_rgb = cv2.cvtColor(original_img, cv2.COLOR_BGR2RGB)
    plt.imshow(original_rgb)
    plt.title("Ảnh gốc (LaPa Aligned)", fontweight='bold')
    plt.axis('off')

    # ---- Hình 2: Mask thô ----
    plt.subplot(1, 3, 2)
    # Dùng bảng màu cmap='tab20' để thấy rõ sự khác biệt giữa 11 classes
    plt.imshow(pred_mask, cmap='tab20', vmin=0, vmax=10)
    plt.title("Dự đoán của MambaVision", fontweight='bold')
    plt.axis('off')

    # ---- Hình 3: Mask đè lên Ảnh gốc ----
    plt.subplot(1, 3, 3)
    plt.imshow(original_rgb)
    # Phủ mask lên trên với độ trong suốt alpha=0.5
    plt.imshow(pred_mask, cmap='tab20', alpha=0.5, vmin=0, vmax=10)
    plt.title("Overlay (Mask + Ảnh gốc)", fontweight='bold')
    plt.axis('off')

    plt.tight_layout()
    plt.savefig('inference_result.png', dpi=300, bbox_inches='tight')
    print("Xong! Hãy mở file 'inference_result.png' bên cột trái để xem thành quả.")

if __name__ == '__main__':
    main()
```

```python
%%writefile test_mamba_mask2former.py
import os
import sys
import torch

os.environ["MPLBACKEND"] = "Agg"
import matplotlib
matplotlib.use("Agg")

# Trỏ vào repo MambaVision
sys.path.insert(0, '/content/MambaVision')
import semantic_segmentation.tools.mamba_vision

from mmengine.config import Config
from mmengine.runner import Runner

def clean_pipeline(pipeline):
    new_pipeline = []
    for t in pipeline:
        if t.get('type') == 'LoadAnnotations':
            t['reduce_zero_label'] = False
        if t.get('type') != 'RandomFlip':
            new_pipeline.append(t)
    return new_pipeline

def main():
    base_config_path = '/content/MambaVision/semantic_segmentation/configs/mask2former/mask2former_swin-s_8xb2-160k_ade20k-512x512.py'

    try:
        cfg = Config.fromfile(base_config_path)
    except FileNotFoundError:
        print("LỖI: Không tìm thấy file base Mask2Former.")
        return

    # ==========================================================
    # 1. ĐỒNG BỘ KIẾN TRÚC MẠNG (Phải khớp 100% với file weights)
    # ==========================================================
    cfg.model.backbone = dict(
        type='MM_mamba_vision',
        depths=(3, 3, 10, 5),
        embed_dims=96,
        dim=128,
        in_dim=64,
        mlp_ratio=4.0,
        num_heads=(2, 4, 8, 16),
        window_size=(8, 8, 64, 32),
        drop_path_rate=0.4,
        layer_scale=1e-05,
        norm_layer='ln2d',
        out_indices=(0, 1, 2, 3),
        init_cfg=None
    )

    mamba_channels = [128, 256, 512, 1024]
    cfg.model.decode_head.in_channels = mamba_channels
    if 'pixel_decoder' in cfg.model.decode_head:
        cfg.model.decode_head.pixel_decoder.in_channels = mamba_channels

    cfg.model.decode_head.num_classes = 11
    cfg.model.decode_head.loss_cls.class_weight = [1.0] * 11 + [0.1]

    # ==========================================================
    # 2. SETUP DATASET CHO CHẾ ĐỘ TEST
    # ==========================================================
    cfg.dataset_type = 'BaseSegDataset'
    cfg.data_root = '/content/data/LaPa_Aligned_512'

    classes = ('background', 'skin', 'left_eyebrow', 'right_eyebrow', 'left_eye',
               'right_eye', 'nose', 'upper_lip', 'inner_mouth', 'lower_lip', 'hair')
    metainfo = dict(classes=classes)

    # Chuyển hướng Test Dataloader vào thư mục TEST
    cfg.test_dataloader.dataset.type = cfg.dataset_type
    cfg.test_dataloader.dataset.data_root = cfg.data_root
    cfg.test_dataloader.dataset.metainfo = metainfo
    cfg.test_dataloader.dataset.reduce_zero_label = False

    # TRỌNG TÂM: Đọc ảnh và nhãn từ thư mục test
    cfg.test_dataloader.dataset.data_prefix = dict(img_path='test/images', seg_map_path='test/labels')

    if 'pipeline' in cfg.test_dataloader.dataset:
        cfg.test_dataloader.dataset.pipeline = clean_pipeline(cfg.test_dataloader.dataset.pipeline)

    # ==========================================================
    # GỌI THẦN CHÚ FULL METRICS: Bật mIoU, Dice, và F-score (sẽ tự suy ra Precision & Recall)
    # ==========================================================
    cfg.test_evaluator = dict(
        type='IoUMetric',
        iou_metrics=['mIoU', 'mDice', 'mFscore']
    )

    # ==========================================================
    # 3. CHỌN FILE WEIGHTS MẠNH NHẤT ĐỂ ĐI THI
    # ==========================================================
    cfg.load_from = '/content/drive/MyDrive/MambaVision/Mask2Former/work_dirs2/phase2/best_mIoU_iter_48000.pth'

    cfg.resume = False
    cfg.work_dir = '/content/drive/MyDrive/MambaVision/Mask2Former/work_dirs2/'

    # 4. KÍCH HOẠT CHẾ ĐỘ LÀM BÀI THI
    runner = Runner.from_cfg(cfg)
    print(f"🚀 BẮT ĐẦU CHẤM ĐIỂM TRÊN TẬP TEST VỚI FILE: {cfg.load_from}")

    runner.test()

if __name__ == '__main__':
    main()
```
