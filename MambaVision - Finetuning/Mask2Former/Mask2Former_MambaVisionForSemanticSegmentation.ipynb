## Phase 1

```python
%%writefile train_mamba_mask2former.py
import os
import sys
import torch

os.environ["MPLBACKEND"] = "Agg"
import matplotlib
matplotlib.use("Agg")

# Trỏ vào repo MambaVision của ông
sys.path.insert(0, '/content/MambaVision')
import semantic_segmentation.tools.mamba_vision

from mmengine.config import Config
from mmengine.runner import Runner

def clean_pipeline(pipeline):
    new_pipeline = []
    for t in pipeline:
        if t.get('type') == 'LoadAnnotations':
            t['reduce_zero_label'] = False
        if t.get('type') != 'RandomFlip':
            new_pipeline.append(t)
    return new_pipeline

def main():
    # 1. LOAD BASE CONFIG MASK2FORMER
    base_config_path = '/content/MambaVision/semantic_segmentation/configs/mask2former/mask2former_swin-s_8xb2-160k_ade20k-512x512.py'

    try:
        cfg = Config.fromfile(base_config_path)
    except FileNotFoundError:
        print("LỖI: Không tìm thấy file base Mask2Former. Hãy chạy lệnh git clone mmseg đã nhé!")
        return

    # 2. ĐỘ KIẾN TRÚC: THAY SWIN BẰNG MAMBAVISION
    cfg.model.backbone = dict(
        type='MM_mamba_vision',
        depths=(3, 3, 10, 5),
        embed_dims=96,
        dim=128,
        in_dim=64,
        mlp_ratio=4.0,
        num_heads=(2, 4, 8, 16),
        window_size=(8, 8, 64, 32),
        drop_path_rate=0.4,
        layer_scale=1e-05,
        norm_layer='ln2d',
        out_indices=(0, 1, 2, 3),
        init_cfg=None # QUAN TRỌNG: Xóa sạch init_cfg để không nạp nhầm weights Swin
    )

    # 3. ĐỒNG BỘ KÊNH (CHANNELS) CHO MASK2FORMER HEAD & PIXEL DECODER
    mamba_channels = [128, 256, 512, 1024]
    cfg.model.decode_head.in_channels = mamba_channels

    # Bắt buộc phải cập nhật cả in_channels của pixel_decoder
    if 'pixel_decoder' in cfg.model.decode_head:
        cfg.model.decode_head.pixel_decoder.in_channels = mamba_channels

    cfg.model.decode_head.num_classes = 11  # LaPa 11 classes

    # VÁ LỖI TENSOR 151: Cập nhật lại mảng class_weight cho hàm Loss
    # 11 số 1.0 cho các class khuôn mặt, và 1 số 0.1 cho class ẩn "no_object"
    cfg.model.decode_head.loss_cls.class_weight = [1.0] * 11 + [0.1]

    # 4. SETUP DATASET LAPA
    cfg.dataset_type = 'BaseSegDataset'
    cfg.data_root = '/content/data/LaPa'

    classes = ('background', 'skin', 'left_eyebrow', 'right_eyebrow', 'left_eye',
               'right_eye', 'nose', 'upper_lip', 'inner_mouth', 'lower_lip', 'hair')
    metainfo = dict(classes=classes)

    for dl in [cfg.train_dataloader, cfg.val_dataloader, cfg.test_dataloader]:
        dl['dataset']['type'] = cfg.dataset_type
        dl['dataset']['data_root'] = cfg.data_root
        dl['dataset']['metainfo'] = metainfo
        dl['dataset']['reduce_zero_label'] = False
        if 'pipeline' in dl['dataset']:
            dl['dataset']['pipeline'] = clean_pipeline(dl['dataset']['pipeline'])

    cfg.train_dataloader.dataset.data_prefix = dict(img_path='train/images', seg_map_path='train/labels')
    cfg.val_dataloader.dataset.data_prefix = dict(img_path='val/images', seg_map_path='val/labels')

    # 5. CHIẾN THUẬT VRAM CHO TESLA T4
    cfg.train_dataloader.batch_size = 2
    cfg.optim_wrapper.accumulative_counts = 8 # Batch ảo = 16

    # ==========================================================
    # 6. NẠP WEIGHTS LAI (HYBRID LOAD)
    # Trỏ ĐÚNG vào file ông vừa phẫu thuật thành công lúc nãy
    # ==========================================================
    # cfg.load_from = '/content/drive/MyDrive/MambaVision/Mask2Former/mamba_backbone_refined.pth'
    cfg.load_from = '/content/drive/MyDrive/MambaVision/Mask2Former/work_dirs/hybrid_run/iter_38000.pth'

    # cfg.resume = False
    cfg.resume = True

    # 7. CHIẾN THUẬT HỌC TẬP (LR phân tầng)
    # cfg.optim_wrapper.optimizer.lr = 1e-4

    cfg.optim_wrapper.paramwise_cfg = dict(
        custom_keys={
            'backbone': dict(lr_mult=0.01, decay_mult=1.0),
            'query_embed': dict(lr_mult=1.0, decay_mult=0.0),
            'query_feat': dict(lr_mult=1.0, decay_mult=0.0),
            'level_embed': dict(lr_mult=1.0, decay_mult=0.0)
        }
    )

    cfg.param_scheduler = [
        dict(
            type='PolyLR',
            eta_min=0,
            power=0.9,
            begin=0,
            end=40000,     # Khớp với max_iters
            by_epoch=False)
    ]

    # 8. CHU KỲ HUẤN LUYỆN
    cfg.train_cfg.max_iters = 40000
    cfg.train_cfg.val_interval = 4000  # [Test] Cứ 4000 vòng mới đem Validation ra chạy 1 lần cho lẹ

    # SETUP HOOK CHECKPOINT THÔNG MINH
    cfg.default_hooks.checkpoint = dict(
        type='CheckpointHook',
        by_epoch=False,
        interval=1000,         # [Backup] Cứ 2000 vòng là tự động lưu file (VD: iter_8000.pth, iter_10000.pth)
        max_keep_ckpts=2,      # Xóa file cũ, chỉ giữ 2 file backup gần nhất để Colab/Drive khỏi bị đầy
        save_best='mIoU',      # [Best] Độc lập với backup! Nó chỉ rình lúc nào Validation chạy xong (mỗi 4k vòng) để xét duyệt mIoU.
        rule='greater',
        save_optimizer=True    # (Tùy chọn) Lưu luôn trạng thái AdamW để resume chuẩn chỉ
    )

    cfg.work_dir = '/content/drive/MyDrive/MambaVision/Mask2Former/work_dirs/hybrid_run'

    runner = Runner.from_cfg(cfg)
    runner.train()

if __name__ == '__main__':
    main()
```

## Inference Test for Face Parsing

```python
%%writefile inference_phase3.py
import os
import sys
import cv2
import torch
import numpy as np

# Tắt hiển thị ngầm để không bị crash trên Colab
if "MPLBACKEND" in os.environ:
    del os.environ["MPLBACKEND"]

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

# Cắm repo MambaVision vào đường dẫn
sys.path.insert(0, '/content/MambaVision')
import semantic_segmentation.tools.mamba_vision

from mmengine.config import Config
from mmseg.apis import init_model, inference_model

def main():
    print("Bắt đầu khởi tạo Mask2Former - MambaVision (Phase 3)...")

    # 1. LOAD BASE CONFIG
    base_config_path = '/content/MambaVision/semantic_segmentation/configs/mask2former/mask2former_swin-s_8xb2-160k_ade20k-512x512.py'

    try:
        cfg = Config.fromfile(base_config_path)
    except FileNotFoundError:
        print(f"Lỗi: Không tìm thấy file {base_config_path}")
        return

    # 2. ĐỒNG BỘ KIẾN TRÚC Y HỆT LÚC TRAIN PHASE 3 (Đã update từ code chuẩn)
    cfg.model.backbone = dict(
        type='MM_mamba_vision',
        depths=(3, 3, 10, 5),
        embed_dims=96,
        dim=128,
        in_dim=64,
        mlp_ratio=4.0,
        num_heads=(2, 4, 8, 16),
        window_size=(8, 8, 64, 32),
        drop_path_rate=0.4,
        layer_scale=1e-05,
        norm_layer='ln2d',
        out_indices=(0, 1, 2, 3),
        init_cfg=None
    )

    mamba_channels = [128, 256, 512, 1024]
    cfg.model.decode_head.in_channels = mamba_channels
    if 'pixel_decoder' in cfg.model.decode_head:
        cfg.model.decode_head.pixel_decoder.in_channels = mamba_channels

    # Định nghĩa classes chuẩn của LaPa
    classes = (
        'background', 'skin', 'left_eyebrow', 'right_eyebrow',
        'left_eye', 'right_eye', 'nose', 'upper_lip',
        'inner_mouth', 'lower_lip', 'hair'
    )
    num_classes = len(classes)
    cfg.model.decode_head.num_classes = num_classes

    # TRIỆT TIÊU PRETRAINED - Tránh lỗi load lộn xộn sinh ra missing keys
    if 'pretrained' in cfg.model.backbone:
        del cfg.model.backbone['pretrained']

    # 3. DỌN DẸP PIPELINE CHUẨN BỊ CHO INFERENCE
    # Fix lại pipeline an toàn nhất để inference_model của MMSeg đọc được mượt
    clean_pipeline = [
        dict(type='LoadImageFromFile'),
        dict(type='Resize', scale=(512, 512), keep_ratio=True),
        dict(type='PackSegInputs')
    ]
    cfg.test_pipeline = clean_pipeline
    if 'test_dataloader' in cfg and 'dataset' in cfg.test_dataloader:
        cfg.test_dataloader.dataset.pipeline = clean_pipeline

    # 4. NẠP WEIGHTS ĐỈNH NHẤT TỪ PHASE 3
    # LƯU Ý: Đảm bảo tên file này chính xác (iter_20000.pth hoặc best_mIoU_iter_20000.pth)
    checkpoint_path = '/content/drive/MyDrive/Pattern Recognition/Source Code/MambaVision - Finetuning/Mask2Former/best_mIoU_iter_20000.pth'

    if not os.path.exists(checkpoint_path):
        print(f"Lỗi: Không tìm thấy file weights tại: {checkpoint_path}")
        return

    # Khởi tạo model và đẩy lên GPU
    model = init_model(cfg, checkpoint_path, device='cuda:0')
    model.dataset_meta = {'classes': classes}

    # 5. CHẠY INFERENCE TRÊN ẢNH THỰC TẾ
    img_path = '/content/data/LaPa/test/images/151362292_2.jpg'

    if not os.path.exists(img_path):
        print(f"Lỗi: Không tìm thấy ảnh test tại: {img_path}")
        return

    try:
        result = inference_model(model, img_path)
    except Exception as e:
        print(f"Lỗi khi chạy Inference: {e}")
        return

    # Lấy ma trận kết quả
    pred_mask = result.pred_sem_seg.data[0].cpu().numpy()

    # 6. VẼ VÀ LƯU KẾT QUẢ CỰC CHẤT LƯỢNG
    plt.figure(figsize=(15, 7))

    # Đọc ảnh gốc bằng OpenCV và đổi hệ màu BGR -> RGB
    original_img = cv2.imread(img_path)
    original_rgb = cv2.cvtColor(original_img, cv2.COLOR_BGR2RGB)

    # Panel 1: Ảnh gốc
    plt.subplot(1, 3, 1)
    plt.imshow(original_rgb)
    plt.title("Ảnh gốc (LaPa)", fontweight='bold')
    plt.axis('off')

    # Panel 2: Bản đồ Mask dự đoán
    plt.subplot(1, 3, 2)
    # Dùng cmap tab20 để ra các mảng màu khác biệt nhau rõ rệt
    plt.imshow(pred_mask, cmap='tab20', vmin=0, vmax=10)
    plt.title("Mặt nạ Dự đoán (Mask2Former)", fontweight='bold')
    plt.axis('off')

    # Panel 3: Overlay đè Mask lên ảnh gốc
    plt.subplot(1, 3, 3)
    plt.imshow(original_rgb)
    # Làm trong suốt cái class Background (0) để lộ cảnh vật gốc
    mask_overlay = np.ma.masked_where(pred_mask == 0, pred_mask)
    plt.imshow(mask_overlay, cmap='tab20', alpha=0.6, vmin=0, vmax=10)
    plt.title("Overlay (Ảnh gốc + Mặt nạ)", fontweight='bold')
    plt.axis('off')

    plt.tight_layout()
    plt.show()
    output_filename = 'inference_phase3_result.png'
    plt.savefig(output_filename, dpi=300, bbox_inches='tight')
    print(f"Đã chạy xong! Hãy mở file '{output_filename}' bên cột trái Colab!")

if __name__ == '__main__':
    main()
```

```python
%%writefile test_mamba_mask2former.py
import os
import sys
import torch

os.environ["MPLBACKEND"] = "Agg"
import matplotlib
matplotlib.use("Agg")

# Trỏ vào repo MambaVision
sys.path.insert(0, '/content/MambaVision')
import semantic_segmentation.tools.mamba_vision

from mmengine.config import Config
from mmengine.runner import Runner

def clean_pipeline(pipeline):
    new_pipeline = []
    for t in pipeline:
        if t.get('type') == 'LoadAnnotations':
            t['reduce_zero_label'] = False
        if t.get('type') != 'RandomFlip':
            new_pipeline.append(t)
    return new_pipeline

def main():
    base_config_path = '/content/MambaVision/semantic_segmentation/configs/mask2former/mask2former_swin-s_8xb2-160k_ade20k-512x512.py'

    try:
        cfg = Config.fromfile(base_config_path)
    except FileNotFoundError:
        print("LỖI: Không tìm thấy file base Mask2Former.")
        return

    # ==========================================================
    # 1. ĐỒNG BỘ KIẾN TRÚC MẠNG (Phải khớp 100% với file weights)
    # ==========================================================
    cfg.model.backbone = dict(
        type='MM_mamba_vision',
        depths=(3, 3, 10, 5),
        embed_dims=96,
        dim=128,
        in_dim=64,
        mlp_ratio=4.0,
        num_heads=(2, 4, 8, 16),
        window_size=(8, 8, 64, 32),
        drop_path_rate=0.4,
        layer_scale=1e-05,
        norm_layer='ln2d',
        out_indices=(0, 1, 2, 3),
        init_cfg=None
    )

    mamba_channels = [128, 256, 512, 1024]
    cfg.model.decode_head.in_channels = mamba_channels
    if 'pixel_decoder' in cfg.model.decode_head:
        cfg.model.decode_head.pixel_decoder.in_channels = mamba_channels

    cfg.model.decode_head.num_classes = 11
    cfg.model.decode_head.loss_cls.class_weight = [1.0] * 11 + [0.1]

    # ==========================================================
    # 2. SETUP DATASET CHO CHẾ ĐỘ TEST
    # ==========================================================
    cfg.dataset_type = 'BaseSegDataset'
    cfg.data_root = '/content/data/LaPa'

    classes = ('background', 'skin', 'left_eyebrow', 'right_eyebrow', 'left_eye',
               'right_eye', 'nose', 'upper_lip', 'inner_mouth', 'lower_lip', 'hair')
    metainfo = dict(classes=classes)

    # Chuyển hướng Test Dataloader vào thư mục TEST
    cfg.test_dataloader.dataset.type = cfg.dataset_type
    cfg.test_dataloader.dataset.data_root = cfg.data_root
    cfg.test_dataloader.dataset.metainfo = metainfo
    cfg.test_dataloader.dataset.reduce_zero_label = False

    # TRỌNG TÂM: Đọc ảnh và nhãn từ thư mục test
    cfg.test_dataloader.dataset.data_prefix = dict(img_path='test/images', seg_map_path='test/labels')

    if 'pipeline' in cfg.test_dataloader.dataset:
        cfg.test_dataloader.dataset.pipeline = clean_pipeline(cfg.test_dataloader.dataset.pipeline)

    # ==========================================================
    # GỌI THẦN CHÚ FULL METRICS: Bật mIoU, Dice, và F-score (sẽ tự suy ra Precision & Recall)
    # ==========================================================
    cfg.test_evaluator = dict(
        type='IoUMetric',
        iou_metrics=['mIoU', 'mDice', 'mFscore']
    )

    # ==========================================================
    # 3. CHỌN FILE WEIGHTS MẠNH NHẤT ĐỂ ĐI THI
    # ==========================================================
    cfg.load_from = '/content/drive/MyDrive/MambaVision/Mask2Former/work_dirs/phase3_20k/best_mIoU_iter_20000.pth'

    cfg.resume = False
    cfg.work_dir = '/content/drive/MyDrive/MambaVision/Mask2Former/work_dirs/test_results'

    # 4. KÍCH HOẠT CHẾ ĐỘ LÀM BÀI THI
    runner = Runner.from_cfg(cfg)
    print(f"🚀 BẮT ĐẦU CHẤM ĐIỂM TRÊN TẬP TEST VỚI FILE: {cfg.load_from}")

    runner.test()

if __name__ == '__main__':
    main()
```
