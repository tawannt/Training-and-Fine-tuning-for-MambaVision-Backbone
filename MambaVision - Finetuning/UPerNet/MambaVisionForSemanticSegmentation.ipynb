## Stage 1

```python
%%writefile train_lapa.py
import os
import sys

os.environ["MPLBACKEND"] = "Agg"
import matplotlib
matplotlib.use("Agg")

sys.path.insert(0, '/content/MambaVision')
import semantic_segmentation.tools.mamba_vision

from mmengine.config import Config
from mmengine.runner import Runner

def main():
    base_config_path = '/content/MambaVision/semantic_segmentation/configs/mamba_vision/mamba_vision_160k_ade20k-512x512_base.py'
    cfg = Config.fromfile(base_config_path)

    cfg.dataset_type = 'BaseSegDataset'
    cfg.data_root = '/content/data/LaPa'

    classes = (
        'background', 'skin', 'left_eyebrow', 'right_eyebrow',
        'left_eye', 'right_eye', 'nose', 'upper_lip',
        'inner_mouth', 'lower_lip', 'hair'
    )
    palette = [
        [0, 0, 0], [204, 0, 0], [76, 153, 0], [204, 204, 0],
        [51, 51, 255], [204, 0, 204], [0, 255, 255], [255, 204, 204],
        [102, 51, 0], [255, 0, 0], [102, 204, 0]
    ]
    metainfo = dict(classes=classes, palette=palette)

    for dl_name in ['train_dataloader', 'val_dataloader', 'test_dataloader']:
        dl = getattr(cfg, dl_name)
        dl['dataset']['type'] = cfg.dataset_type
        dl['dataset']['data_root'] = cfg.data_root
        dl['dataset']['metainfo'] = metainfo
        dl['dataset']['reduce_zero_label'] = False
        dl['dataset'].pop('classes', None)
        dl['dataset'].pop('palette', None)

        if 'pipeline' in dl['dataset']:
            for t in dl['dataset']['pipeline']:
                if t.get('type') == 'LoadAnnotations':
                    t['reduce_zero_label'] = False

    # BẮT BUỘC: ÉP BATCH SIZE LỚN HƠN 1 ĐỂ TRỊ BỆNH BATCH NORM
    cfg.train_dataloader.batch_size = 4
    cfg.val_dataloader.batch_size = 1 # Val không cần tính gradient, để 1 cho nhẹ VRAM

    cfg.train_dataloader.dataset.data_prefix = dict(img_path='train/images', seg_map_path='train/labels')
    cfg.train_dataloader.dataset.seg_map_suffix = '.png'

    cfg.val_dataloader.dataset.data_prefix = dict(img_path='val/images', seg_map_path='val/labels')
    cfg.val_dataloader.dataset.seg_map_suffix = '.png'

    cfg.test_dataloader.dataset.data_prefix = dict(img_path='test/images', seg_map_path='test/labels')

    for p_name in ['train_pipeline', 'val_pipeline', 'test_pipeline']:
        if hasattr(cfg, p_name):
            for t in getattr(cfg, p_name):
                if t.get('type') == 'LoadAnnotations':
                    t['reduce_zero_label'] = False

    if hasattr(cfg, 'tta_pipeline'):
        for t in cfg.tta_pipeline:
            if 'transforms' in t:
                for sub_list in t['transforms']:
                    for sub_t in sub_list:
                        if sub_t.get('type') == 'LoadAnnotations':
                            sub_t['reduce_zero_label'] = False

    num_classes = len(classes)
    cfg.model.decode_head.num_classes = num_classes
    cfg.model.auxiliary_head.num_classes = num_classes

    if 'pretrained' in cfg.model.backbone:
        del cfg.model.backbone['pretrained']

    cfg.load_from = '/content/mamba_vision_160k_ade20k-512x512_base.pth'
    cfg.optim_wrapper.optimizer.lr = 0.00006
    cfg.train_cfg.max_iters = 20000
    cfg.train_cfg.val_interval = 2000
    cfg.default_hooks.checkpoint.interval = 2000
    cfg.work_dir = './drive/MyDrive/MambaVision/work_dirs/lapa_mambavision'

    runner = Runner.from_cfg(cfg)
    runner.train()

if __name__ == '__main__':
    main()
```

### Resume training with Gradient accummulation from 1 checkpoint

```python
%%writefile train_lapa_2.py
import os
import sys

os.environ["MPLBACKEND"] = "Agg"
import matplotlib
matplotlib.use("Agg")

sys.path.insert(0, '/content/MambaVision')
import semantic_segmentation.tools.mamba_vision

from mmengine.config import Config
from mmengine.runner import Runner

# Hàm dọn dẹp Pipeline: Ép reduce_zero_label = False và XÓA RandomFlip
def clean_pipeline(pipeline):
    new_pipeline = []
    for t in pipeline:
        if t.get('type') == 'LoadAnnotations':
            t['reduce_zero_label'] = False
        if t.get('type') != 'RandomFlip':  # TIÊU DIỆT RANDOM FLIP
            new_pipeline.append(t)
    return new_pipeline

def main():
    base_config_path = '/content/MambaVision/semantic_segmentation/configs/mamba_vision/mamba_vision_160k_ade20k-512x512_base.py'
    cfg = Config.fromfile(base_config_path)

    cfg.dataset_type = 'BaseSegDataset'
    cfg.data_root = '/content/data/LaPa'

    classes = (
        'background', 'skin', 'left_eyebrow', 'right_eyebrow',
        'left_eye', 'right_eye', 'nose', 'upper_lip',
        'inner_mouth', 'lower_lip', 'hair'
    )
    palette = [
        [0, 0, 0], [204, 0, 0], [76, 153, 0], [204, 204, 0],
        [51, 51, 255], [204, 0, 204], [0, 255, 255], [255, 204, 204],
        [102, 51, 0], [255, 0, 0], [102, 204, 0]
    ]
    metainfo = dict(classes=classes, palette=palette)

    for dl_name in ['train_dataloader', 'val_dataloader', 'test_dataloader']:
        dl = getattr(cfg, dl_name)
        dl['dataset']['type'] = cfg.dataset_type
        dl['dataset']['data_root'] = cfg.data_root
        dl['dataset']['metainfo'] = metainfo
        dl['dataset']['reduce_zero_label'] = False
        dl['dataset'].pop('classes', None)
        dl['dataset'].pop('palette', None)

        # Dọn dẹp pipeline lồng bên trong dataset
        if 'pipeline' in dl['dataset']:
            dl['dataset']['pipeline'] = clean_pipeline(dl['dataset']['pipeline'])

    cfg.train_dataloader.num_workers = 2
    cfg.val_dataloader.num_workers = 2
    cfg.test_dataloader.num_workers = 2

    # Giữ nguyên cấu hình ép xung x16
    cfg.train_dataloader.batch_size = 4
    cfg.val_dataloader.batch_size = 1
    cfg.optim_wrapper.accumulative_counts = 4

    cfg.train_dataloader.dataset.data_prefix = dict(img_path='train/images', seg_map_path='train/labels')
    cfg.train_dataloader.dataset.seg_map_suffix = '.png'
    cfg.val_dataloader.dataset.data_prefix = dict(img_path='val/images', seg_map_path='val/labels')
    cfg.val_dataloader.dataset.seg_map_suffix = '.png'
    cfg.test_dataloader.dataset.data_prefix = dict(img_path='test/images', seg_map_path='test/labels')

    # Dọn dẹp pipeline toàn cục
    for p_name in ['train_pipeline', 'val_pipeline', 'test_pipeline']:
        if hasattr(cfg, p_name):
            setattr(cfg, p_name, clean_pipeline(getattr(cfg, p_name)))

    num_classes = len(classes)
    cfg.model.decode_head.num_classes = num_classes
    cfg.model.auxiliary_head.num_classes = num_classes

    if 'pretrained' in cfg.model.backbone:
        del cfg.model.backbone['pretrained']

    # Lùi lại mốc 2000 (mốc chưa bị hỏng spatial features nghiêm trọng)
    cfg.load_from = './drive/MyDrive/MambaVision/work_dirs/lapa_mambavision/iter_17000.pth'
    cfg.resume = True
    # cfg.optim_wrapper.optimizer.lr = 0.0001

    cfg.train_cfg.max_iters = 20000
    cfg.train_cfg.val_interval = 1000
    # cfg.default_hooks.checkpoint.interval = 1000
    cfg.default_hooks.checkpoint = dict(
        type='CheckpointHook',
        by_epoch=False,
        interval=1000,       # Vẫn lưu định kỳ mỗi 1000 bước để phòng hờ máy sập
        save_best='mIoU',    # [QUAN TRỌNG] Kích hoạt lưu file best dựa trên điểm mIoU
        rule='greater',      # Quy tắc: mIoU càng lớn càng tốt
        max_keep_ckpts=3     # Tiết kiệm dung lượng: Chỉ giữ lại tối đa 3 file checkpoint định kỳ mới nhất
    )
    cfg.work_dir = './drive/MyDrive/MambaVision/work_dirs/lapa_mambavision'

    runner = Runner.from_cfg(cfg)
    runner.train()

if __name__ == '__main__':
    main()
```

```python
%%writefile train_lapa_stage2.py
import os
import sys

os.environ["MPLBACKEND"] = "Agg"
import matplotlib
matplotlib.use("Agg")

sys.path.insert(0, '/content/MambaVision')
import semantic_segmentation.tools.mamba_vision

from mmengine.config import Config
from mmengine.runner import Runner

def clean_pipeline(pipeline):
    new_pipeline = []
    for t in pipeline:
        if t.get('type') == 'LoadAnnotations':
            t['reduce_zero_label'] = False
        if t.get('type') != 'RandomFlip':
            new_pipeline.append(t)
    return new_pipeline

def main():
    base_config_path = '/content/MambaVision/semantic_segmentation/configs/mamba_vision/mamba_vision_160k_ade20k-512x512_base.py'
    cfg = Config.fromfile(base_config_path)

    cfg.dataset_type = 'BaseSegDataset'
    cfg.data_root = '/content/data/LaPa'

    classes = (
        'background', 'skin', 'left_eyebrow', 'right_eyebrow',
        'left_eye', 'right_eye', 'nose', 'upper_lip',
        'inner_mouth', 'lower_lip', 'hair'
    )
    palette = [
        [0, 0, 0], [204, 0, 0], [76, 153, 0], [204, 204, 0],
        [51, 51, 255], [204, 0, 204], [0, 255, 255], [255, 204, 204],
        [102, 51, 0], [255, 0, 0], [102, 204, 0]
    ]
    metainfo = dict(classes=classes, palette=palette)

    for dl_name in ['train_dataloader', 'val_dataloader', 'test_dataloader']:
        dl = getattr(cfg, dl_name)
        dl['dataset']['type'] = cfg.dataset_type
        dl['dataset']['data_root'] = cfg.data_root
        dl['dataset']['metainfo'] = metainfo
        dl['dataset']['reduce_zero_label'] = False
        dl['dataset'].pop('classes', None)
        dl['dataset'].pop('palette', None)

        if 'pipeline' in dl['dataset']:
            dl['dataset']['pipeline'] = clean_pipeline(dl['dataset']['pipeline'])

    cfg.train_dataloader.num_workers = 2
    cfg.val_dataloader.num_workers = 2
    cfg.test_dataloader.num_workers = 2

    cfg.train_dataloader.batch_size = 4
    cfg.val_dataloader.batch_size = 1
    cfg.optim_wrapper.accumulative_counts = 4

    cfg.train_dataloader.dataset.data_prefix = dict(img_path='train/images', seg_map_path='train/labels')
    cfg.train_dataloader.dataset.seg_map_suffix = '.png'
    cfg.val_dataloader.dataset.data_prefix = dict(img_path='val/images', seg_map_path='val/labels')
    cfg.val_dataloader.dataset.seg_map_suffix = '.png'
    cfg.test_dataloader.dataset.data_prefix = dict(img_path='test/images', seg_map_path='test/labels')

    for p_name in ['train_pipeline', 'val_pipeline', 'test_pipeline']:
        if hasattr(cfg, p_name):
            setattr(cfg, p_name, clean_pipeline(getattr(cfg, p_name)))

    num_classes = len(classes)
    cfg.model.decode_head.num_classes = num_classes
    cfg.model.auxiliary_head.num_classes = num_classes

    if 'pretrained' in cfg.model.backbone:
        del cfg.model.backbone['pretrained']

    # =================================================================
    # VŨ KHÍ GỌT VIỀN: HYBRID LOSS (CE + DICE)
    # =================================================================
    cfg.model.decode_head.loss_decode = [
        dict(type='CrossEntropyLoss', use_sigmoid=False, loss_weight=1.0, loss_name='loss_ce'),
        dict(type='DiceLoss', loss_weight=1.5, loss_name='loss_dice') # Phạt nặng ranh giới sai
    ]
    cfg.model.auxiliary_head.loss_decode = [
        dict(type='CrossEntropyLoss', use_sigmoid=False, loss_weight=0.4, loss_name='loss_ce'),
        dict(type='DiceLoss', loss_weight=0.6, loss_name='loss_dice')
    ]

    # =================================================================
    # CHIẾN THUẬT STAGE 2 FINE-TUNING
    # =================================================================
    # 1. Bốc đúng con best của chặng 1 làm vạch xuất phát
    cfg.load_from = './drive/MyDrive/MambaVision/work_dirs/lapa_mambavision/best_mIoU_iter_20000.pth'

    # 2. BẮT BUỘC FALSE: Để reset trí nhớ của AdamW, tránh sốc gradient với hàm Loss mới
    cfg.resume = False

    # 3. Ép Learning Rate cực nhỏ bằng cách ghi đè trực tiếp
    cfg.optim_wrapper.optimizer.lr = 0.00001 # 1e-5 (Dao mổ y tế)

    # 4. Tắt PolyLR (không cho LR tụt nữa, giữ nguyên 1e-5 chạy rà viền)
    cfg.param_scheduler = []

    # Cấu hình chu kỳ ngắn
    cfg.train_cfg.max_iters = 10000
    cfg.train_cfg.val_interval = 1000

    # Tiếp tục cơ chế Save Best để lỡ mô hình gãy giữa chừng thì vẫn còn vớt được đỉnh
    cfg.default_hooks.checkpoint = dict(
        type='CheckpointHook',
        by_epoch=False,
        interval=1000,
        save_best='mIoU',
        rule='greater',
        max_keep_ckpts=2
    )

    cfg.work_dir = './drive/MyDrive/MambaVision/work_dirs/lapa_stage2_hybrid'

    runner = Runner.from_cfg(cfg)
    runner.train()

if __name__ == '__main__':
    main()
```

## Stage 3

```python
%%writefile resume_lapa_stage3.py
import os
import sys
import torch
import torch.nn as nn
import torch.nn.functional as F

os.environ["MPLBACKEND"] = "Agg"
import matplotlib
matplotlib.use("Agg")

sys.path.insert(0, '/content/MambaVision')
import semantic_segmentation.tools.mamba_vision

from mmengine.config import Config
from mmengine.runner import Runner
from mmseg.registry import MODELS
from mmseg.models.losses import DiceLoss

# =================================================================
# GIẢI PHÁP ĐẶC TRỊ BUG CỦA MMSEGMENTATION (Phần 1: CE)
# =================================================================
@MODELS.register_module()
class SafeWeightedCE(nn.Module):
    def __init__(self, class_weight=None, loss_weight=1.0, ignore_index=255, loss_name='loss_ce'):
        super().__init__()
        self.class_weight = class_weight
        self.loss_weight = loss_weight
        self.ignore_index = ignore_index
        self._loss_name = loss_name

    def forward(self, cls_score, label, weight=None, **kwargs):
        if label.dim() == 4:
            label = label.squeeze(1)

        loss = F.cross_entropy(cls_score, label, reduction='none', ignore_index=self.ignore_index)
        valid_mask = (label != self.ignore_index)

        if self.class_weight is not None:
            cw_tensor = cls_score.new_tensor(self.class_weight)
            safe_label = torch.where(valid_mask, label, torch.zeros_like(label))
            pixel_cw = cw_tensor[safe_label]
            loss = loss * pixel_cw

        divisor = valid_mask.sum().float() + 1e-6
        if weight is not None:
            weight = weight.float()
            loss = loss * weight
            divisor = weight.sum() + 1e-6

        loss = loss.sum() / divisor
        return self.loss_weight * loss

    @property
    def loss_name(self):
        return self._loss_name

# =================================================================
# GIẢI PHÁP ĐẶC TRỊ BUG CỦA MMSEGMENTATION (Phần 2: Dice)
# Chặn đứng ma trận OHEM không cho tràn vào Dice Loss
# =================================================================
@MODELS.register_module()
class SafeDiceLoss(nn.Module):
    def __init__(self, **kwargs):
        super().__init__()
        self.dice_loss = DiceLoss(**kwargs)

    def forward(self, cls_score, label, weight=None, **kwargs):
        # TUYỆT ĐỐI KHÔNG truyền 'weight' (từ OHEM) vào hàm gốc
        return self.dice_loss(cls_score, label, weight=None, **kwargs)

    @property
    def loss_name(self):
        return self.dice_loss.loss_name

def clean_pipeline(pipeline):
    new_pipeline = []
    for t in pipeline:
        if t.get('type') == 'LoadAnnotations':
            t['reduce_zero_label'] = False
        if t.get('type') != 'RandomFlip':
            new_pipeline.append(t)
    return new_pipeline

def main():
    base_config_path = '/content/MambaVision/semantic_segmentation/configs/mamba_vision/mamba_vision_160k_ade20k-512x512_base.py'
    cfg = Config.fromfile(base_config_path)

    cfg.dataset_type = 'BaseSegDataset'
    cfg.data_root = '/content/data/LaPa'

    classes = (
        'background', 'skin', 'left_eyebrow', 'right_eyebrow',
        'left_eye', 'right_eye', 'nose', 'upper_lip',
        'inner_mouth', 'lower_lip', 'hair'
    )
    palette = [
        [0, 0, 0], [204, 0, 0], [76, 153, 0], [204, 204, 0],
        [51, 51, 255], [204, 0, 204], [0, 255, 255], [255, 204, 204],
        [102, 51, 0], [255, 0, 0], [102, 204, 0]
    ]
    metainfo = dict(classes=classes, palette=palette)

    for dl_name in ['train_dataloader', 'val_dataloader', 'test_dataloader']:
        dl = getattr(cfg, dl_name)
        dl['dataset']['type'] = cfg.dataset_type
        dl['dataset']['data_root'] = cfg.data_root
        dl['dataset']['metainfo'] = metainfo
        dl['dataset']['reduce_zero_label'] = False
        dl['dataset'].pop('classes', None)
        dl['dataset'].pop('palette', None)

        if 'pipeline' in dl['dataset']:
            dl['dataset']['pipeline'] = clean_pipeline(dl['dataset']['pipeline'])

    cfg.train_dataloader.num_workers = 2
    cfg.val_dataloader.num_workers = 2
    cfg.test_dataloader.num_workers = 2

    cfg.train_dataloader.batch_size = 4
    cfg.val_dataloader.batch_size = 1
    cfg.optim_wrapper.accumulative_counts = 4

    cfg.train_dataloader.dataset.data_prefix = dict(img_path='train/images', seg_map_path='train/labels')
    cfg.train_dataloader.dataset.seg_map_suffix = '.png'
    cfg.val_dataloader.dataset.data_prefix = dict(img_path='val/images', seg_map_path='val/labels')
    cfg.val_dataloader.dataset.seg_map_suffix = '.png'
    cfg.test_dataloader.dataset.data_prefix = dict(img_path='test/images', seg_map_path='test/labels')

    for p_name in ['train_pipeline', 'val_pipeline', 'test_pipeline']:
        if hasattr(cfg, p_name):
            setattr(cfg, p_name, clean_pipeline(getattr(cfg, p_name)))

    num_classes = len(classes)
    cfg.model.decode_head.num_classes = num_classes
    cfg.model.auxiliary_head.num_classes = num_classes

    if 'pretrained' in cfg.model.backbone:
        del cfg.model.backbone['pretrained']

    # =================================================================
    # SỬ DỤNG SAFE_WEIGHTED_CE VÀ SAFE_DICE KẾT HỢP OHEM
    # =================================================================
    # stage3_class_weights = [0.1, 0.2, 1.5, 1.5, 1.2, 1.2, 0.8, 2.5, 2.0, 2.5, 0.2]
    # Trọng số mềm mỏng hơn. Không đè bẹp Skin/Bg xuống 0.1 nữa để nó giữ được ranh giới.
    stage3_class_weights = [0.8, 0.8, 1.2, 1.2, 1.1, 1.1, 0.9, 1.5, 1.5, 1.5, 0.8]

    # cfg.model.decode_head.sampler = dict(type='OHEMPixelSampler', thresh=0.9, min_kept=131072)
    cfg.model.decode_head.sampler = dict(type='OHEMPixelSampler', thresh=0.9, min_kept=32768)

    cfg.model.decode_head.loss_decode = [
        dict(
            type='SafeWeightedCE',
            class_weight=stage3_class_weights,
            # loss_weight=1.0,
            loss_weight=0.5,
            loss_name='loss_ce_ohem'
        ),
        dict(
            type='SafeDiceLoss',
            # loss_weight=0.5,
            loss_weight=1.5,
            loss_name='loss_dice') # Gọi SafeDiceLoss
    ]

    if hasattr(cfg.model, 'auxiliary_head') and cfg.model.auxiliary_head is not None:
        # cfg.model.auxiliary_head.sampler = dict(type='OHEMPixelSampler', thresh=0.9, min_kept=131072)
        cfg.model.auxiliary_head.sampler = dict(type='OHEMPixelSampler', thresh=0.9, min_kept=32768)

        cfg.model.auxiliary_head.loss_decode = [
            dict(
                type='SafeWeightedCE',
                class_weight=stage3_class_weights,
                # loss_weight=0.4,
                loss_weight=0.2,
                loss_name='loss_ce_ohem'
            ),
            # dict(type='SafeDiceLoss', loss_weight=0.2, loss_name='loss_dice') # Gọi SafeDiceLoss
            dict(type='SafeDiceLoss', loss_weight=0.6, loss_name='loss_dice') # Gọi SafeDiceLoss
        ]

    # =================================================================
    # CHIẾN THUẬT HUẤN LUYỆN
    # =================================================================
    cfg.optim_wrapper.paramwise_cfg = dict(
        custom_keys={
            'backbone': dict(lr_mult=0.0, decay_mult=0.0)
        }
    )

    cfg.load_from = './drive/MyDrive/MambaVision/work_dirs/lapa_stage2_hybrid/best_mIoU_iter_9000.pth'
    cfg.resume = False
    cfg.optim_wrapper.optimizer.lr = 5e-5

    cfg.param_scheduler = [
        dict(
            type='CosineAnnealingLR',
            T_max=5000,
            eta_min=1e-6,
            begin=0,
            end=5000,
            by_epoch=False
        )
    ]

    cfg.train_cfg.max_iters = 5000
    cfg.train_cfg.val_interval = 1000

    cfg.default_hooks.checkpoint = dict(
        type='CheckpointHook',
        by_epoch=False,
        interval=1000,
        save_best='mIoU',
        rule='greater',
        max_keep_ckpts=2
    )

    cfg.work_dir = './drive/MyDrive/MambaVision/work_dirs/lapa_stage3_ohem2'

    runner = Runner.from_cfg(cfg)
    runner.train()

if __name__ == '__main__':
    main()
```
